# Feature Types

```python
cat_cols = X.select_dtypes(include=['object', 'category']).columns.to_list()
num_cols = X.select_dtypes(include=[np.number]).columns.to_list()
```

```python
print("Categorical columns : ", cat_cols)
print("Numerical columns : ", num_cols)
```

# preprocessing pipeline

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
```

```python
numerical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler', StandardScaler())
])
```

```python
categorical_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])
```

```python
preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_transformer, num_cols),
    ('cat', categorical_transformer, cat_cols)
], remainder='drop')
```

```python
preprocessor
```

# Stage - 1 Classification baseline

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report
```

```python
X.head()
```

```python
y = df_clf[target_col].astype(str).str.strip().str.lower()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=10, stratify=y)
print("Train size : ", X_train.shape)
print("Test size : ", X_test.shape)
```

# Fit RF baseline pipelines

```python
rf_clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('rf_cls', RandomForestClassifier())
])
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
param_grid = {
    'rf_cls__n_estimators' : [100,200,300,400],
    'rf_cls__max_depth' : [None, 4, 8, 10],
    'rf_cls__max_features' : ['sqrt', None]
}

grid = GridSearchCV(rf_clf_pipeline, param_grid, cv=5, scoring='f1', n_jobs=-1)
grid.fit(X_train, y_train)
print("Best params (rf_cls) : ", grid.best_params_)
best_clf = grid.best_estimator_
y_pred_best = best_clf.predict(X_test)
print(classification_report(y_test, y_pred_best))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, roc_auc_score
import matplotlib.pyplot as plt

# Predicted probabilities for ROC-AUC
y_prob = best_clf.predict_proba(X_test)[:, 1]

# ROC-AUC
print("ROC-AUC:", roc_auc_score(y_test, y_prob))

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred_best)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=best_clf.classes_
)

disp.plot()
plt.title("Stage 1 - Confusion Matrix")
plt.show()
```

```python
# Random forest final baseline model
rf_clf_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('rf_cls', RandomForestClassifier(n_estimators=400, max_depth=None, max_features=None,random_state=10, oob_score=True))
])

rf_clf_pipeline.fit(X_train, y_train)
y_pred_cls = rf_clf_pipeline.predict(X_test)
print(classification_report(y_test, y_pred_cls))
print("RF cls OOB score :", getattr(rf_clf_pipeline.named_steps['rf_cls'], 'oob_score_', None))
```

```python
rf_clf_pipeline
```

# checking feature importance

```python
feature_importance = rf_clf_pipeline.named_steps['rf_cls'].feature_importances_

preprocessor = rf_clf_pipeline.named_steps['preprocessor']

feature_names = preprocessor.get_feature_names_out()

importance_df = pd.DataFrame(
    {
        'Feature' : feature_names,
        'Importance'  : feature_importance
    }
)

importance_df = importance_df.sort_values(by='Importance', ascending=False)

print("Top 10 Important Features : \n", importance_df.head(10))

plt.figure(figsize=(10,6))
plt.barh(importance_df['Feature'][:10], importance_df['Importance'][:10])
plt.xlabel("Random forest feature importance")
plt.ylabel("Feature")
plt.title("Top 10 Important Features")
plt.show()
```

```python
reg_target = "loan_amount"
reg_df = approved_df.dropna(subset=[reg_target]).copy()
```

```python
X_reg = reg_df.drop(columns=[reg_target, "loan_status"])
y_reg = reg_df[reg_target]
```

```python
cat_cols_reg = X_reg.select_dtypes(include=['object', 'category']).columns.to_list()
num_cols_reg = X_reg.select_dtypes(include=[np.number]).columns.to_list()
```

```python
print("Categorical columns : ", cat_cols_reg)
print("Numerical columns : ", num_cols_reg)
```

```python
Xr_train, Xr_test, yr_train, yr_test = train_test_split(X_reg, y_reg, test_size=0.2, random_state=10)
print("Train size : ", Xr_train.shape)
print("Test size : ", Xr_test.shape)
```

```python
reg_preprocessor = ColumnTransformer(transformers=[
    ('num', numerical_transformer, num_cols_reg),
    ('cat', categorical_transformer, cat_cols_reg)
], remainder='drop')
```

# Fit RF_reg baseline model

```python
from sklearn.ensemble import RandomForestRegressor
```

```python
rf_reg_pipeline = Pipeline(steps=[
    ('preprocessor', reg_preprocessor),
    ('rf_reg', RandomForestRegressor())
])
```

```python
param_reg_grid = {
    'rf_reg__n_estimators' : [100,200,300,400],
    'rf_reg__max_depth' : [None, 4, 8, 10],
    'rf_reg__max_features' : ['sqrt', None],
    'rf_reg__min_samples_split' : [2,5,10]
}
```

```python
grid = GridSearchCV(rf_reg_pipeline, param_reg_grid, cv=5, scoring='r2', n_jobs=-1)
grid.fit(Xr_train, yr_train)
```

```python
print("Best params (rf_reg) : ", grid.best_params_)
best_reg = grid.best_estimator_
```

```python
grid.best_score_
```

# Prediction Function

```python
clf = joblib.load("../models/stage_1_rf_classifier_pipeline.pkl")
reg = joblib.load("../models/stage_2_rf_regression_pipeline.pkl")
```

```python
# 100 columns
# 10 columns

def two_stage_predict(applicant_df):
    out = {}
    applicant_df = applicant_df.copy()
    # stage 1
    approve = clf.predict(applicant_df)[0]
    # print(approve)
    out['loan_status'] = int(approve)

    if approve == 1:
        # stage 2
        loan_amount_pred = reg.predict(applicant_df)[0]
        out['regression_prediction'] = float(loan_amount_pred)
    else:
        print("Not Approved")
    
    return out
```

# SHAP EXPLANABILITY

```python
import shap
import matplotlib.pyplot as plt

# Extract components from the trained pipeline
preprocessor = best_clf.named_steps['preprocessor']
rf_model = best_clf.named_steps['rf_cls']

# Transform test data
X_test_transformed = preprocessor.transform(X_test)

print("Transformed shape:", X_test_transformed.shape)
```

```python
explainer = shap.TreeExplainer(rf_model)

shap_values = explainer(X_test_transformed)

print(shap_values.shape)
```

```python
# SHAP values for the approved class (class 1)
shap_values_class1 = shap_values.values[:, :, 1]

print(shap_values_class1.shape)
```

```python
feature_names = preprocessor.get_feature_names_out()

print(feature_names)
```

```python
shap.summary_plot(
    shap_values_class1,
    X_test_transformed,
    feature_names=feature_names
)
```

```python
preprocessor_reg = best_reg.named_steps['preprocessor']
rf_reg_model = best_reg.named_steps['rf_reg']

Xr_test_transformed = preprocessor_reg.transform(Xr_test)

print("Transformed shape:", Xr_test_transformed.shape)
```

```python
explainer_reg = shap.TreeExplainer(rf_reg_model)

shap_values_reg = explainer_reg(Xr_test_transformed)

print(shap_values_reg.shape)
```

```python
feature_names_reg = preprocessor_reg.get_feature_names_out()

print(feature_names_reg)
print(len(feature_names_reg))
```

```python
shap.summary_plot(
    shap_values_reg,
    Xr_test_transformed,
    feature_names=feature_names_reg
)
```

# SHAP ANALYSIS CONCLUSION
- CIBIL score is the most influential feature for loan approval predictions.
- Annual income is the dominant feature for predicting the loan amount.
- Asset-related features such as bank, residential, luxury, and commercial assets have moderate influence.
- Education and self-employment have relatively low impact on the model predictions.
- SHAP analysis improves model interpretability by showing how individual features influence predictions.
